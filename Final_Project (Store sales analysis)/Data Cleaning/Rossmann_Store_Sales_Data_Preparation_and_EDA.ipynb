```python
from google.colab import files

uploaded = files.upload()
```

```python
# ROSSMANN STORE SALES
# DATA PREPARATION, DATA CLEANING & EXPLORATORY DATA ANALYSIS

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import io
import os
import zipfile
import calendar
from pathlib import Path
from IPython.display import display

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully!")


# 1. IDENTIFY AND LOAD THE THREE DATASETS

datasets = {}

for filename, content in uploaded.items():

    if not filename.lower().endswith(".csv"):
        continue

    preview = pd.read_csv(
        io.BytesIO(content),
        nrows=5,
        low_memory=False
    )

    columns = set(preview.columns.str.strip())

    if {"Sales", "Customers", "Store"}.issubset(columns):
        dataset_name = "train"

    elif {"Id", "Store", "Date"}.issubset(columns) and "Sales" not in columns:
        dataset_name = "test"

    elif {"StoreType", "Assortment", "Promo2"}.issubset(columns):
        dataset_name = "store"

    else:
        raise ValueError(
            f"Cannot identify dataset: {filename}"
        )

    if dataset_name in datasets:
        raise ValueError(
            f"More than one {dataset_name} file was uploaded. "
            "Upload only one copy of each dataset."
        )

    datasets[dataset_name] = pd.read_csv(
        io.BytesIO(content),
        low_memory=False
    )

required = {"train", "test", "store"}

if set(datasets.keys()) != required:
    raise ValueError(
        "Please upload exactly these three datasets: "
        "train.csv, test.csv, and store.csv."
    )

train = datasets["train"]
test = datasets["test"]
store = datasets["store"]

print("\nAll three datasets loaded successfully!")

print("Train:", train.shape)
print("Test:", test.shape)
print("Store:", store.shape)


# 2. INITIAL DATA INSPECTION

print("\nTRAIN SAMPLE")
display(train.head())

print("\nTEST SAMPLE")
display(test.head())

print("\nSTORE SAMPLE")
display(store.head())

print("\nDATASET INFORMATION")

for name, df in [
    ("TRAIN", train),
    ("TEST", test),
    ("STORE", store)
]:
    print(f"\n{name}")
    print("Shape:", df.shape)
    print("Duplicate rows:", df.duplicated().sum())
    display(df.isnull().sum().to_frame("Missing Values"))


# 3. CREATE PROJECT OUTPUT FOLDERS

output_dir = Path("Rossmann_Submission")

figures_dir = output_dir / "figures"
reports_dir = output_dir / "reports"

figures_dir.mkdir(parents=True, exist_ok=True)
reports_dir.mkdir(parents=True, exist_ok=True)


# 4. CLEAN TRAIN DATASET

train_cleaned = train.copy()

train_cleaned.columns = (
    train_cleaned.columns.str.strip()
)

train_cleaned["Date"] = pd.to_datetime(
    train_cleaned["Date"],
    errors="coerce"
)

if train_cleaned["Date"].isna().any():
    raise ValueError("Invalid dates found in Train.")

train_integer_columns = [
    "Store",
    "DayOfWeek",
    "Sales",
    "Customers",
    "Open",
    "Promo",
    "SchoolHoliday"
]

for col in train_integer_columns:

    train_cleaned[col] = pd.to_numeric(
        train_cleaned[col],
        errors="coerce"
    )

    if train_cleaned[col].isna().any():
        raise ValueError(
            f"Unexpected missing or invalid values in Train: {col}"
        )

    train_cleaned[col] = train_cleaned[col].astype("int64")

train_cleaned["StateHoliday"] = (
    train_cleaned["StateHoliday"]
    .astype("string")
    .str.strip()
)

# Date-based features
train_cleaned["Year"] = train_cleaned["Date"].dt.year
train_cleaned["Month"] = train_cleaned["Date"].dt.month
train_cleaned["Day"] = train_cleaned["Date"].dt.day

# Remove exact duplicate records only
train_cleaned = train_cleaned.drop_duplicates()

print("\nTrain cleaning completed!")
print("Final Train shape:", train_cleaned.shape)


# 5. CLEAN TEST DATASET

test_cleaned = test.copy()

test_cleaned.columns = (
    test_cleaned.columns.str.strip()
)

test_cleaned["Date"] = pd.to_datetime(
    test_cleaned["Date"],
    errors="coerce"
)

if test_cleaned["Date"].isna().any():
    raise ValueError("Invalid dates found in Test.")

test_integer_columns = [
    "Id",
    "Store",
    "DayOfWeek",
    "Promo",
    "SchoolHoliday"
]

for col in test_integer_columns:

    test_cleaned[col] = pd.to_numeric(
        test_cleaned[col],
        errors="coerce"
    )

    if test_cleaned[col].isna().any():
        raise ValueError(
            f"Unexpected missing or invalid values in Test: {col}"
        )

    test_cleaned[col] = test_cleaned[col].astype("int64")

# Open contains 11 genuine missing values
test_cleaned["Open"] = pd.to_numeric(
    test_cleaned["Open"],
    errors="coerce"
).astype("Int64")

test_cleaned["OpenMissing"] = (
    test_cleaned["Open"].isna().astype("int8")
)

test_cleaned["StateHoliday"] = (
    test_cleaned["StateHoliday"]
    .astype("string")
    .str.strip()
)

test_cleaned["Year"] = test_cleaned["Date"].dt.year
test_cleaned["Month"] = test_cleaned["Date"].dt.month
test_cleaned["Day"] = test_cleaned["Date"].dt.day

test_cleaned = test_cleaned.drop_duplicates()

print("\nTest cleaning completed!")
print("Final Test shape:", test_cleaned.shape)


# 6. CLEAN STORE DATASET

store_cleaned = store.copy()

store_cleaned.columns = (
    store_cleaned.columns.str.strip()
)

store_numeric_columns = [
    "Store",
    "CompetitionDistance",
    "CompetitionOpenSinceMonth",
    "CompetitionOpenSinceYear",
    "Promo2",
    "Promo2SinceWeek",
    "Promo2SinceYear"
]

for col in store_numeric_columns:

    store_cleaned[col] = pd.to_numeric(
        store_cleaned[col],
        errors="coerce"
    )

integer_store_columns = [
    "Store",
    "CompetitionOpenSinceMonth",
    "CompetitionOpenSinceYear",
    "Promo2SinceWeek",
    "Promo2SinceYear",
    "Promo2"
]

for col in integer_store_columns:
    store_cleaned[col] = store_cleaned[col].astype("Int64")

text_columns = [
    "StoreType",
    "Assortment",
    "PromoInterval"
]

for col in text_columns:

    store_cleaned[col] = (
        store_cleaned[col]
        .astype("string")
        .str.strip()
    )

# Normalize September abbreviation
store_cleaned["PromoInterval"] = (
    store_cleaned["PromoInterval"]
    .str.replace("Sept", "Sep", regex=False)
)

# Missing-value indicators
store_cleaned["CompetitionDistanceMissing"] = (
    store_cleaned["CompetitionDistance"]
    .isna()
    .astype("int8")
)

store_cleaned["CompetitionOpenDateMissing"] = (
    store_cleaned["CompetitionOpenSinceYear"]
    .isna()
    .astype("int8")
)

store_cleaned["Promo2InfoMissing"] = (
    store_cleaned["Promo2SinceYear"]
    .isna()
    .astype("int8")
)

# Competition distance categories
store_cleaned["CompetitionDistanceCategory"] = pd.cut(
    store_cleaned["CompetitionDistance"],
    bins=[-np.inf, 1000, 5000, np.inf],
    labels=["Very Close", "Medium", "Far"]
)

store_cleaned = store_cleaned.drop_duplicates()

print("\nStore cleaning completed!")
print("Final Store shape:", store_cleaned.shape)


# 7. DATA VALIDATION

print("\n========== DATA VALIDATION ==========")

print("Train rows:", len(train_cleaned))
print("Test rows:", len(test_cleaned))
print("Store rows:", len(store_cleaned))

print(
    "Train duplicate Store-Date combinations:",
    train_cleaned.duplicated(["Store", "Date"]).sum()
)

print(
    "Test duplicate Store-Date combinations:",
    test_cleaned.duplicated(["Store", "Date"]).sum()
)

print(
    "Duplicate Store IDs:",
    store_cleaned["Store"].duplicated().sum()
)

# Verify all Store IDs exist in Store dataset
train_unmatched = (
    set(train_cleaned["Store"])
    - set(store_cleaned["Store"])
)

test_unmatched = (
    set(test_cleaned["Store"])
    - set(store_cleaned["Store"])
)

print("Unmatched Train Store IDs:", len(train_unmatched))
print("Unmatched Test Store IDs:", len(test_unmatched))

assert len(train_unmatched) == 0
assert len(test_unmatched) == 0

assert train_cleaned.duplicated(["Store", "Date"]).sum() == 0
assert test_cleaned.duplicated(["Store", "Date"]).sum() == 0

assert store_cleaned["Store"].is_unique

assert len(train_cleaned) == len(train)
assert len(test_cleaned) == len(test)
assert len(store_cleaned) == len(store)

print("\nAll validation checks passed!")


# 8. MERGE TRAIN AND TEST WITH STORE INFORMATION

train_merged = train_cleaned.merge(
    store_cleaned,
    on="Store",
    how="left",
    validate="many_to_one"
)

test_merged = test_cleaned.merge(
    store_cleaned,
    on="Store",
    how="left",
    validate="many_to_one"
)

assert len(train_merged) == len(train_cleaned)
assert len(test_merged) == len(test_cleaned)

print("\nMerge completed successfully!")
print("Train merged:", train_merged.shape)
print("Test merged:", test_merged.shape)



# 9. FEATURE ENGINEERING

def add_business_features(df, store_dimension):

    df = df.copy()

    #Competition opening duration in months

    competition_months = (
        (df["Date"].dt.year - df["CompetitionOpenSinceYear"]) * 12
        + (
            df["Date"].dt.month
            - df["CompetitionOpenSinceMonth"]
        )
    )

    df["CompetitionOpenMonths"] = (
        competition_months.astype("Int64")
    )

    # Calculate the actual Promo2 start date
    # using ISO year and week

    promo_lookup = store_dimension.loc[
        store_dimension["Promo2"].eq(1),
        [
            "Store",
            "Promo2SinceYear",
            "Promo2SinceWeek",
            "PromoInterval"
        ]
    ].copy()

    promo_year = (
        promo_lookup["Promo2SinceYear"]
        .astype("Int64")
        .astype("string")
    )

    promo_week = (
        promo_lookup["Promo2SinceWeek"]
        .astype("Int64")
        .astype("string")
        .str.zfill(2)
    )

    promo_lookup["Promo2StartDate"] = pd.to_datetime(
        promo_year + "-W" + promo_week + "-1",
        format="%G-W%V-%u",
        errors="coerce"
    )

    if promo_lookup["Promo2StartDate"].isna().any():
        raise ValueError("Invalid Promo2 start dates found.")

    #Expand PromoInterval into individual months

    promo_lookup["PromoMonth"] = (
        promo_lookup["PromoInterval"]
        .str.replace("Sept", "Sep", regex=False)
        .str.split(",")
    )

    promo_lookup = promo_lookup.explode("PromoMonth")

    promo_lookup["PromoMonth"] = (
        promo_lookup["PromoMonth"].str.strip()
    )

    promo_lookup["PromoMonthNumber"] = (
        pd.to_datetime(
            promo_lookup["PromoMonth"],
            format="%b",
            errors="coerce"
        ).dt.month
    )

    if promo_lookup["PromoMonthNumber"].isna().any():
        raise ValueError("Invalid PromoInterval month found.")

    # Check both the active month and actual start date

    df["IsPromo2Month"] = np.zeros(
        len(df),
        dtype="int8"
    )

    for month_number in range(1, 13):

        active_stores = (
            promo_lookup.loc[
                promo_lookup["PromoMonthNumber"] == month_number,
                ["Store", "Promo2StartDate"]
            ]
            .set_index("Store")["Promo2StartDate"]
        )

        store_start_dates = df["Store"].map(active_stores)

        active_mask = (
            (df["Date"].dt.month == month_number)
            & store_start_dates.notna()
            & (df["Date"] >= store_start_dates)
        )

        df.loc[active_mask, "IsPromo2Month"] = 1

    return df


train_merged = add_business_features(
    train_merged,
    store_cleaned
)

test_merged = add_business_features(
    test_merged,
    store_cleaned
)

print("Feature engineering completed successfully!")

print("\nPromo2 feature distribution in Train:")
print(train_merged["IsPromo2Month"].value_counts())

print("\nPromo2 feature distribution in Test:")
print(test_merged["IsPromo2Month"].value_counts())


# 10. PREPARE DATA FOR SALES ANALYSIS

sales_data = train_merged.loc[
    train_merged["Open"] == 1
].copy()

print("Open store records:", len(sales_data))


# 11. KPI SUMMARY

total_sales = sales_data["Sales"].sum()
total_customers = sales_data["Customers"].sum()

kpi_summary = pd.DataFrame({
    "Metric": [
        "Total Stores",
        "Total Sales",
        "Average Sales per Open Store-Day",
        "Total Customers",
        "Average Customers per Open Store-Day",
        "Average Sales per Customer",
        "Open Store-Day Records",
        "Promotion Records"
    ],
    "Value": [
        train_cleaned["Store"].nunique(),
        total_sales,
        sales_data["Sales"].mean(),
        total_customers,
        sales_data["Customers"].mean(),
        total_sales / total_customers if total_customers else np.nan,
        len(sales_data),
        sales_data["Promo"].sum()
    ]
})

display(kpi_summary)

kpi_summary.to_csv(
    reports_dir / "kpi_summary.csv",
    index=False
)


# 12. MONTHLY SALES TREND

monthly_sales = (
    sales_data
    .groupby(sales_data["Date"].dt.to_period("M"))["Sales"]
    .mean()
)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Average Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Average Sales per Open Store-Day")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    figures_dir / "monthly_sales_trend.png",
    dpi=150
)

plt.show()


# 13. SALES BY DAY OF WEEK

day_sales = (
    sales_data
    .groupby("DayOfWeek")["Sales"]
    .mean()
)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=day_sales.index,
    y=day_sales.values
)

plt.title("Average Sales by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Average Sales")

plt.tight_layout()

plt.savefig(
    figures_dir / "sales_by_day_of_week.png",
    dpi=150
)

plt.show()


# 14. PROMOTION ANALYSIS

promo_sales = (
    sales_data
    .groupby("Promo")
    .agg(
        AverageSales=("Sales", "mean"),
        AverageCustomers=("Customers", "mean"),
        Records=("Sales", "count")
    )
    .reset_index()
)

display(promo_sales)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=promo_sales,
    x="Promo",
    y="AverageSales"
)

plt.title("Average Sales With and Without Promotion")
plt.xlabel("Promo (0 = No, 1 = Yes)")
plt.ylabel("Average Sales")

plt.tight_layout()

plt.savefig(
    figures_dir / "promo_sales.png",
    dpi=150
)

plt.show()


# ============================================================
# 15. STORE TYPE ANALYSIS
# ============================================================

store_type_sales = (
    sales_data
    .groupby("StoreType")
    .agg(
        AverageSales=("Sales", "mean"),
        TotalSales=("Sales", "sum"),
        AverageCustomers=("Customers", "mean"),
        StoreCount=("Store", "nunique")
    )
    .reset_index()
)

display(store_type_sales)

store_type_sales.to_csv(
    reports_dir / "store_type_analysis.csv",
    index=False
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=store_type_sales,
    x="StoreType",
    y="AverageSales"
)

plt.title("Average Sales by Store Type")
plt.xlabel("Store Type")
plt.ylabel("Average Sales")

plt.tight_layout()

plt.savefig(
    figures_dir / "sales_by_store_type.png",
    dpi=150
)

plt.show()


# 16. CUSTOMERS VS SALES

sample_data = sales_data.sample(
    n=min(10000, len(sales_data)),
    random_state=42
)

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=sample_data,
    x="Customers",
    y="Sales",
    alpha=0.4
)

plt.title("Relationship Between Customers and Sales")
plt.xlabel("Customers")
plt.ylabel("Sales")

plt.tight_layout()

plt.savefig(
    figures_dir / "customers_vs_sales.png",
    dpi=150
)

plt.show()


# 17. COMPETITION DISTANCE ANALYSIS

competition_analysis = (
    sales_data
    .groupby(
        "CompetitionDistanceCategory",
        observed=True
    )
    .agg(
        AverageSales=("Sales", "mean"),
        AverageCustomers=("Customers", "mean"),
        StoreCount=("Store", "nunique")
    )
    .reset_index()
)

display(competition_analysis)

competition_analysis.to_csv(
    reports_dir / "competition_analysis.csv",
    index=False
)


# 18. STATE HOLIDAY ANALYSIS

holiday_sales = (
    sales_data
    .groupby("StateHoliday")
    .agg(
        AverageSales=("Sales", "mean"),
        AverageCustomers=("Customers", "mean"),
        Records=("Sales", "count")
    )
    .reset_index()
)

display(holiday_sales)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=holiday_sales,
    x="StateHoliday",
    y="AverageSales"
)

plt.title("Average Sales by State Holiday")
plt.xlabel("State Holiday")
plt.ylabel("Average Sales")

plt.tight_layout()

plt.savefig(
    figures_dir / "state_holiday_sales.png",
    dpi=150
)

plt.show()


# 19. ASSORTMENT ANALYSIS

assortment_analysis = (
    sales_data
    .groupby("Assortment")
    .agg(
        AverageSales=("Sales", "mean"),
        AverageCustomers=("Customers", "mean"),
        StoreCount=("Store", "nunique")
    )
    .reset_index()
)

display(assortment_analysis)

assortment_analysis.to_csv(
    reports_dir / "assortment_analysis.csv",
    index=False
)


# 20. DATA QUALITY REPORT

quality_reports = []

for name, df in [
    ("Train", train_cleaned),
    ("Test", test_cleaned),
    ("Store", store_cleaned)
]:

    for col in df.columns:

        quality_reports.append({
            "Dataset": name,
            "Column": col,
            "DataType": str(df[col].dtype),
            "MissingValues": int(df[col].isna().sum()),
            "MissingPercentage": round(
                df[col].isna().mean() * 100,
                2
            ),
            "UniqueValues": int(df[col].nunique())
        })

quality_report = pd.DataFrame(quality_reports)

quality_report.to_csv(
    reports_dir / "data_quality_report.csv",
    index=False
)

display(quality_report.head(20))


# 21. VALIDATION SUMMARY

validation_summary = pd.DataFrame({
    "Check": [
        "Train row count preserved",
        "Test row count preserved",
        "Store row count preserved",
        "Train Store-Date duplicates",
        "Test Store-Date duplicates",
        "Duplicate Store IDs",
        "Unmatched Train Store IDs",
        "Unmatched Test Store IDs",
        "Missing Test Open values"
    ],
    "Result": [
        len(train_cleaned) == len(train),
        len(test_cleaned) == len(test),
        len(store_cleaned) == len(store),
        train_cleaned.duplicated(["Store", "Date"]).sum(),
        test_cleaned.duplicated(["Store", "Date"]).sum(),
        store_cleaned["Store"].duplicated().sum(),
        len(train_unmatched),
        len(test_unmatched),
        test_cleaned["Open"].isna().sum()
    ]
})

display(validation_summary)

validation_summary.to_csv(
    reports_dir / "validation_summary.csv",
    index=False
)


# 22. EXPORT CLEANED DATASETS

train_cleaned.to_csv(
    output_dir / "train_cleaned.csv",
    index=False
)

test_cleaned.to_csv(
    output_dir / "test_cleaned.csv",
    index=False
)

store_cleaned.to_csv(
    output_dir / "store_cleaned.csv",
    index=False
)

train_merged.to_csv(
    output_dir / "train_merged.csv",
    index=False
)

test_merged.to_csv(
    output_dir / "test_merged.csv",
    index=False
)

print("\nAll datasets exported successfully!")


# 23. GENERATE README AUTOMATICALLY

readme = f"""
# Rossmann Store Sales
## Data Preparation and Exploratory Data Analysis

### Project Overview

This project focuses on data preparation, cleaning,
feature engineering, and exploratory data analysis
of the Rossmann Store Sales dataset.

### Source Datasets

- Train: {train.shape[0]:,} rows and {train.shape[1]} columns.
- Test: {test.shape[0]:,} rows and {test.shape[1]} columns.
- Store: {store.shape[0]:,} rows and {store.shape[1]} columns.

### Data Preparation

- Standardized column names and data types.
- Converted date columns.
- Checked duplicate records.
- Added missing-value indicators.
- Preserved structurally missing Promo2 information.
- Preserved unknown competition information.
- Created competition and promotion features.
- Merged sales data with store information.
- Validated row counts and Store relationships.

### Exploratory Data Analysis

The analysis includes:

- Monthly sales trends.
- Average sales by day of week.
- Promotion analysis.
- Store type comparison.
- Customers vs. sales.
- Competition distance analysis.
- State holiday analysis.
- Assortment analysis.

### Important Data Notes

- Missing Open values in Test were preserved and flagged.
- Missing Promo2 values for stores not enrolled in Promo2
  were preserved.
- Missing competition information was not artificially imputed.
- Outliers were not automatically removed.
- Descriptive associations do not establish causation.

### Output Files

- train_cleaned.csv
- test_cleaned.csv
- store_cleaned.csv
- train_merged.csv
- test_merged.csv
- figures/
- reports/

### Tools

Python, Pandas, NumPy, Matplotlib, Seaborn, Google Colab.
"""

with open(
    output_dir / "README.md",
    "w",
    encoding="utf-8"
) as file:
    file.write(readme)


# 24. FINAL PROJECT SUMMARY

print("\n========== FINAL PROJECT SUMMARY ==========")

print("Train:", train_cleaned.shape)
print("Test:", test_cleaned.shape)
print("Store:", store_cleaned.shape)

print("\nOutput files:")

for path in output_dir.rglob("*"):

    if path.is_file():
        print(path)


# 25. CREATE ZIP FILE

zip_name = "Rossmann_Data_Preparation_and_EDA.zip"

with zipfile.ZipFile(
    zip_name,
    "w",
    zipfile.ZIP_DEFLATED
) as zip_file:

    for path in output_dir.rglob("*"):

        if path.is_file():

            zip_file.write(
                path,
                arcname=str(path)
            )

print("\nZIP created successfully!")

print(
    "ZIP size in MB:",
    round(os.path.getsize(zip_name) / (1024 * 1024), 2)
)


# 26. DOWNLOAD FINAL PROJECT FILES

files.download(zip_name)

print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
